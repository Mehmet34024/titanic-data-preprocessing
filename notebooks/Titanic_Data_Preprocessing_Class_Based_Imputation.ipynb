```python
import pandas as pd
import numpy as np

# Veriyi okuyoruz
df = pd.read_csv('train.csv')

# ilk 5 satiri kontrol ediyoruz
df.head()
```

```python
# Her sütundaki toplam eksik değer sayısını ekrana yazdıralım
print("--- Sütunlardaki Eksik Değer Sayıları ---")
print(df.isnull().sum())
```

```python
# Pclass (1, 2, 3) sınıflarına göre Age sütununun ortalamasını hesaplayıp görelim
class_age_means = df.groupby('Pclass')['Age'].mean()

print("--- Bilet Sınıfına (Pclass) Göre Yaş Ortalamarı ---")
print(class_age_means.round(2))
```

```python
# Age sütunundaki boşlukları ait olduğu Pclass grubunun yaş ortalaması ile dolduruyoruz
df['Age'] = df['Age'].fillna(df.groupby('Pclass')['Age'].transform('mean'))
```

```python
# 1. Kontrol: Age sütununda kaç tane eksik değer kaldı? (0 görmeliyiz)
print(f"Age sütunundaki kalan eksik değer sayısı: {df['Age'].isnull().sum()}")

# 2. İlk 15 satırı ekrana basalım
df.head(15)
```
